```python
import pandas as pd

url = "https://raw.githubusercontent.com/hokuamadata/spacex-falcon9-landing-prediction/main/data/raw/spacex_launch_dash.csv"
df = pd.read_csv(url)

# 発射場ごと
print(df.groupby("Launch Site")["class"].agg(successes="sum", n="size", rate="mean"))

# 重さの範囲ごと（ダッシュボードで読んだ 2000 kg 区切りに合わせる）
bins = pd.cut(df["Payload Mass (kg)"], [0, 2000, 4000, 6000, 8000, 10000], include_lowest=True)
print(df.groupby(bins, observed=True)["class"].agg(n="size", successes="sum", rate="mean"))

# ブースターの型ごと
print(df.groupby("Booster Version Category")["class"].agg(n="size", successes="sum", rate="mean"))
```

#### Finding Insights through Visual Analytics

*Note: `class` indicates whether the first stage **landed** successfully, so "launch success" in the questions is read as "landing success".*
*Note: Payload ranges were compared in 2,000 kg bins.*

- **Which site has the most successful landings?** → KSC LC-39A (41.7% of all successes)
- **Which site has the highest landing success rate?** → KSC LC-39A (76.9%, 10 of 13 flights). However, KSC was only used from 2017 onward, after the technology had matured, so the difference is not due to the site alone.
- **Which payload range has the highest landing success rate?** → 2,000–4,000 kg (about 62%, 13 of 21 flights)
- **Which payload range has the lowest landing success rate?** → 6,000–8,000 kg (0%, 0 of 4 flights). However, no landing was attempted on any of these 4 flights (expendable boosters), so they are not landing failures. Payloads of 8,000 kg or more succeeded on 3 of 5 flights.
- **Which booster version has the highest landing success rate?** → B5 has a 100% success rate, but with only one flight, no conclusion can be drawn. Among versions with multiple flights, FT has the highest rate (about 65%, 15 of 23 flights).

#### 視覚分析で洞察を見つける

※ class は第1段の**着陸**の成否なので、問いの「打ち上げ成功」は「着陸成功」と読み替えている。

※ 重さの範囲は 2,000 kg ごとに区切って比べた。

- **着陸成功が最も多い発射場は？** → KSC LC-39A（全成功の41.7%）
- **着陸成功率が最も高い発射場は？** → KSC LC-39A（76.9%、13便中10便）。ただし KSC は2017年以降しか使われておらず、技術が成熟したあとの便が中心のため、発射場そのものの差だけとは言えない。
- **着陸成功率が最も高い重さの範囲は？** → 2,000〜4,000 kg（約62%、21便中13便）
- **着陸成功率が最も低い重さの範囲は？** → 6,000〜8,000 kg（0%、4便中0便）。ただし4便とも着陸を試みていない便（使い捨て）で、着陸に失敗したわけではない。なお 8,000 kg 以上は5便中3便が成功している。
- **着陸成功率が最も高いブースターの型は？** → 成功率100%は B5 だが、1便しかないため判断できない。複数の便がある型の中では、FT が最も高い（約65%、23便中15便）。
